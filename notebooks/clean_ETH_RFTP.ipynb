```python
cols_to_keep=  ['country','adm1_name','mkt_name', 'lat', 'lon',
    'price_date', 'currency',
    'maize', 'sorghum', 'teff_fao', 'wheat', 'livestock_goat'
]
```

```python
df1=df[cols_to_keep].copy()
```

```python
df1_long=df1.melt(id_vars=['country','adm1_name','mkt_name', 'lat', 'lon', 'price_date', 'currency'],
                  value_vars=['maize', 'sorghum', 'teff_fao', 'wheat', 'livestock_goat'],
                  var_name='product', value_name='value')
```

```python
df1_long['product']=df1_long['product'].replace({ 'teff_fao': 'teff' })
```

```python
df1_long=df1_long.dropna(subset=['value'])
```

```python
# Rename columns so they match Dataset 2
df1_long = df1_long.rename(columns={
    'adm1_name': 'admin_1',
    'mkt_name': 'market',
    'lat': 'latitude',
    'lon': 'longitude',
    'price_date': 'period_date'
})
```

```python
print("shape")
print(df1_long.shape)
print("their info")
print(df1_long.info())
print("theri desciption")
print(df1_long.describe(include='all'))
print("is it null")
print(df1_long.isnull().sum())
print(df1_long.head(20))
```

```python
# Fill missing lat/lon using the same market's known coordinates
df1_long['latitude'] = df1_long.groupby('market')['latitude'].transform(lambda x: x.fillna(x.median()))
df1_long['longitude'] = df1_long.groupby('market')['longitude'].transform(lambda x: x.fillna(x.median()))
```

```python
print(df1_long.isnull().sum())
print(df1_long.duplicated().sum())
```

```python
# Show the rows that still have missing latitude
missing_coords = df1_long[df1_long['latitude'].isna()]

print("Number of rows with missing coordinates:", len(missing_coords))
print("\nMarkets with missing coordinates:")
print(missing_coords['market'].value_counts())

print("\nAdmin_1 of those markets:")
print(missing_coords[['admin_1', 'market']].drop_duplicates())
```

```python
# Remove the fake "Market Average" rows
df1_long = df1_long[df1_long['market'] != 'Market Average']
```

```python
print("Shape after removing Market Average:", df1_long.shape)
print("\nMissing values:")
print(df1_long.isnull().sum())
print("\nDuplicates:", df1_long.duplicated().sum())
print("\nUnique markets:", df1_long['market'].nunique())
```

```python
import os

# Create the directory if it doesn't exist
os.makedirs("../data/processed", exist_ok=True)

# Now save the file
df1_long.to_csv("../data/processed/ETH_RTFP_cleaned_2007_2026-07-06.csv", index=False)
```
